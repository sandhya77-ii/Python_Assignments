```python
#Task1
name = input("Enter your name: ")
total_spent = 0
```

```python
# Task 2: Track spending for 3 days
for day in range(1, 4):
    print("\nDay", day)

    spent = float(input("Enter money spent today ($): "))

    # Adding daily spending to the total
    total_spent += spent

    # Checking the daily budget
    if spent > 20:
        print("Over your $20 budget today!")
    else:
        print("Great! Under budget today.")
```

```python
# Task 3: Display the final summary
print("\n----- Final Summary -----")
print("Name:", name)
print("Total money spent over 3 days: $", total_spent)

# Check the total budget
if total_spent <= 60:
    print("Overall Result: You stayed under your $60 total budget!")
else:
    print("Overall Result: You went over your $60 total budget!")
```
