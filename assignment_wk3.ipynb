```python
# Task 1: Creating inventory and order queue

# Creating the inventory dictionary with available stock counts
inventory = {
    "Widget": 10,
    "Gadget": 5,
    "Sensor": 0,
    "Cable": 15
}

# Creating the order queue as a list of lists
order_queue = [
    ["Widget", 3],
    ["Sensor", 2],
    ["Gadget", 7],
    ["Cable", 10],
    ["Mouse", 4]
]

print("Initial Inventory:", inventory)
print("Order Queue:", order_queue)
```

```python
# Task 2: Processing incoming orders

# Creating a list for storing orders that are not fully fulfilled
unfulfilled_orders = []

# Creating a variable for counting fully fulfilled orders
fully_fulfilled_orders = 0

# Iterating through each order in the order queue
for order in order_queue:
    item = order[0]
    requested_quantity = order[1]

    # Looking up the requested item in the inventory dictionary
    stock = inventory.get(item)

    # Checking whether the item is missing from the inventory
    if stock is None:
        print("Missing Item Alert:", item, "is not available in inventory.")
        unfulfilled_orders.append([item, requested_quantity])

    # Checking whether the item is out of stock
    elif stock == 0:
        print("Out-of-Stock Alert:", item, "has no stock available.")
        unfulfilled_orders.append([item, requested_quantity])

    # Checking whether enough stock is available for full fulfillment
    elif stock >= requested_quantity:
        inventory[item] = stock - requested_quantity
        fully_fulfilled_orders += 1
        print("Order Successfully Fulfilled:", item, "-", requested_quantity, "units")

    # Processing the order when only part of the requested stock is available
    elif stock > 0 and stock < requested_quantity:
        unfulfilled_quantity = requested_quantity - stock
        inventory[item] = 0
        unfulfilled_orders.append([item, unfulfilled_quantity])
        print(
            "Order Partially Fulfilled:",
            item,
            "- Supplied:",
            stock,
            "units, Unfulfilled:",
            unfulfilled_quantity,
            "units"
        )
```

```python
# Task 3: Displaying analytics and summary output

# Printing the final updated inventory dictionary
print("\n--- End-of-Batch Summary ---")
print("Final Updated Inventory:", inventory)

# Printing the total number of fully fulfilled orders
print("Total Fully Fulfilled Orders:", fully_fulfilled_orders)

# Printing the list of items and quantities that could not be fully supplied
print("Unfulfilled Items/Quantities:", unfulfilled_orders)
```
