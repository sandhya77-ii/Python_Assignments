```python
# Task 1: Open the input and output files
input_file = open("raw_logs.txt", "r")
master_file = open("decrypted_master.txt", "w")
alert_file = open("security_alerts.txt", "w")

# Task 2: Reading and decrypting each line
for encrypted_line in input_file:
    decrypted_line = ""

    for char in encrypted_line:
        if "A" <= char <= "Z":
            decrypted_char = chr(
                (ord(char) - ord("A") - 3) % 26 + ord("A")
            )
            decrypted_line += decrypted_char

        elif "a" <= char <= "z":
            decrypted_char = chr(
                (ord(char) - ord("a") - 3) % 26 + ord("a")
            )
            decrypted_line += decrypted_char

        else:
            decrypted_line += char

#Task 3: Write the decrypted line to the master file and check for critical alerts
    # Remove leading and trailing whitespace
    decrypted_line = decrypted_line.strip()

    # Write every decrypted line to the master file
    master_file.write(decrypted_line + "\n")

    # Write critical alerts to the alert file
    if "BREACH" in decrypted_line.upper():
        alert_file.write(decrypted_line + "\n")

# Close all files
input_file.close()
master_file.close()
alert_file.close()

print("Log decryption completed successfully!")
print("Decrypted logs saved in decrypted_master.txt")
print("Security alerts saved in security_alerts.txt")
```
