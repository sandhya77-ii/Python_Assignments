# Task 1: Functional tools (filter, map, lambda)

```python
def clean_transactions(amounts, min_threshold=10.0):
    """Keep only positive amounts that are at least min_threshold."""
    return list(filter(lambda amt: amt > 0 and amt >= min_threshold, amounts))


def apply_service_fee(amounts, fee_rate=0.02):
    """Deduct a service fee (fee_rate) from every amount."""
    return list(map(lambda amt: round(amt * (1 - fee_rate), 2), amounts))


# Test cases for Task 1
raw_batch = [5, -20, 0, 10, 15.5, 250, 9.99, 100]

print("Raw batch           :", raw_batch)
print("Clean (default 10)  :", clean_transactions(raw_batch))
print("Clean (threshold 50):", clean_transactions(raw_batch, min_threshold=50))
print("Clean (empty list)  :", clean_transactions([]))

print()
print("Fee 2% (default)    :", apply_service_fee([100, 200, 50]))
print("Fee 5%              :", apply_service_fee([100, 200, 50], fee_rate=0.05))
print("Fee on cleaned batch:", apply_service_fee(clean_transactions(raw_batch)))
```

# Task 2: Purely recursive nested total

```python
def calculate_nested_total(data):
    """Return the total value of a number or an arbitrarily nested list."""
    # Base case 1: empty list
    if data == []:
        return 0.0
    # Base case 2: a single number
    if isinstance(data, (int, float)):
        return data
    # Recursive step: split into head and tail, solve both, add them
    if isinstance(data, list):
        head = data[0]
        tail = data[1:]
        return calculate_nested_total(head) + calculate_nested_total(tail)
    raise TypeError("Only numbers and lists are allowed")


# ---- Test cases for Task 2 ----
print("Sample batch      :", calculate_nested_total([100, [50, 25], [[10, 5], 40]]))
print("Empty list        :", calculate_nested_total([]))
print("Single number     :", calculate_nested_total(75.5))
print("Flat list         :", calculate_nested_total([10, 20, 30]))
print("Only empty lists  :", calculate_nested_total([[], [[]], []]))
print("Deep nesting      :", calculate_nested_total([1, [2, [3, [4, [5]]]]]))
print("Floats mixed      :", calculate_nested_total([[10.5, 4.5], 5, [[0.25]]]))
```

# Combined audit pipeline (Task 1 + Task 2)

```python
# Flat batch -> clean -> deduct fee -> nested sub-transaction audit
batch = [5, -20, 0, 10, 15.5, 250, 9.99, 100]
valid = clean_transactions(batch)
processed = apply_service_fee(valid)
print("Valid transactions  :", valid)
print("After 2% fee        :", processed)
print("Total (flat)        :", round(calculate_nested_total(processed), 2))

# Nested sub-transactions triggered by high-value operations
nested_batch = [250, [100, [50, 25]], [[10, 5], 40]]
print("Nested audit total  :", calculate_nested_total(nested_batch))
```
